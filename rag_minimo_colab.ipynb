# RAG Mínimo

Proyecto de Retrieval-Augmented Generation listo para **Google Colab**.

**Pregunta:** ¿cómo recuperar documentos relevantes y generar respuestas?

**Método:** Embeddings + similitud coseno + prompt simple

**Stack:** sentence-transformers, sklearn

Documentos de ejemplo.

## 0. Setup

```python
!pip install -q sentence-transformers

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
print("OK")
```

## 1. Documentos de ejemplo

```python
documentos = [
    "El Titanic se hundió en 1912.",
    "Python es un lenguaje de programación popular para ciencia de datos.",
    "La inteligencia artificial incluye machine learning y deep learning.",
    "Google Colab permite ejecutar notebooks en la nube.",
    "Los embeddings representan texto como vectores."
]
print(f"Documentos: {len(documentos)}")
```

## 2. Embeddings

```python
modelo = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = modelo.encode(documentos)
print(embeddings.shape)
```

## 3. Recuperación

```python
def recuperar(query, top_k=2):
    query_emb = modelo.encode([query])
    sim = cosine_similarity(query_emb, embeddings)[0]
    indices = np.argsort(sim)[::-1][:top_k]
    return [documentos[i] for i in indices]

print(recuperar("¿Qué es Python?"))
```

## 4. Generación simple (prompt)

```python
def responder(query):
    contexto = recuperar(query)
    prompt = f"Contexto: {' '.join(contexto)}\nPregunta: {query}\nRespuesta:"
    return prompt  # en producción pasar a un LLM

print(responder("¿Qué es Python?"))
```

## 5. Conclusiones

- RAG combina recuperación de documentos con generación.
- Ideal para chatbots con conocimiento propio.
- Se puede extender a FAISS, Chroma o LLMs reales (Gemini, GPT).

**Consejo:** usá un vector store para bases grandes.
